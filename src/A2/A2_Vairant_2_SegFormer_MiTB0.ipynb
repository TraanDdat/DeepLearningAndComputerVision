```python
import os
from pathlib import Path
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from transformers import SegformerModel, SegformerConfig
from torchvision.datasets import Cityscapes
from torchvision.transforms import functional as TF
from torchvision.transforms import ColorJitter, InterpolationMode, RandomApply
from PIL import Image
from tqdm import tqdm
```

# Pre-Configuration

```python
# Dataset
DATA_ROOT = "/content/data/data"
CACHE_ROOT = "/content/cache"

NUM_CLASSES = 19
IGNORE_INDEX = 255

# Input
IMAGE_HEIGHT = 512
IMAGE_WIDTH = 1024

# Training
BATCH_SIZE = 2
EPOCHS = 10
VAL_INTERVAL = 2
BACKBONE_LR = 6e-5
HEAD_LR = 6e-4
WEIGHT_DECAY = 0.01
NUM_WORKERS = 2
PREFETCH_FACTOR = 2

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
if DEVICE.type == "cuda":
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

print("Device:", DEVICE)
```

# Dataset Processing

```python
ID_TO_TRAIN_ID = np.full(256, IGNORE_INDEX, dtype=np.uint8)

for original_id, train_id in {
    7: 0, 8: 1, 11: 2, 12: 3, 13: 4,
    17: 5, 19: 6, 20: 7, 21: 8, 22: 9,
    23: 10, 24: 11, 25: 12, 26: 13, 27: 14,
    28: 15, 31: 16, 32: 17, 33: 18
}.items():
    ID_TO_TRAIN_ID[original_id] = train_id


def build_cache(split):
    """Resize + remap Avoid decode on each Epoch"""

    image_dir = Path(CACHE_ROOT) / split / "images"
    mask_dir = Path(CACHE_ROOT) / split / "masks"

    image_dir.mkdir(parents=True, exist_ok=True)
    mask_dir.mkdir(parents=True, exist_ok=True)

    source = Cityscapes(
        root=DATA_ROOT,
        split=split,
        mode="fine",
        target_type="semantic"
    )

    for index in tqdm(range(len(source)), desc=f"Caching {split}"):

        image_path = image_dir / f"{index:05d}.jpg"
        mask_path = mask_dir / f"{index:05d}.png"

        if image_path.exists() and mask_path.exists():
            continue

        image, mask = source[index]

        image.resize(
            (IMAGE_WIDTH, IMAGE_HEIGHT),
            Image.BILINEAR
        ).save(image_path, quality=95)

        mask = mask.resize(
            (IMAGE_WIDTH, IMAGE_HEIGHT),
            Image.NEAREST
        )

        Image.fromarray(
            ID_TO_TRAIN_ID[np.array(mask, dtype=np.uint8)]
        ).save(mask_path)


build_cache("train")
build_cache("val")
```

```python
CLASS_NAMES = [
    "road",
    "sidewalk",
    "building",
    "wall",
    "fence",
    "pole",
    "traffic light",
    "traffic sign",
    "vegetation",
    "terrain",
    "sky",
    "person",
    "rider",
    "car",
    "truck",
    "bus",
    "train",
    "motorcycle",
    "bicycle"
]


class CityscapesDataset(torch.utils.data.Dataset):

    def __init__(self, split="train", augment=False):

        self.image_dir = Path(CACHE_ROOT) / split / "images"
        self.mask_dir = Path(CACHE_ROOT) / split / "masks"

        self.names = sorted(
            path.stem for path in self.image_dir.glob("*.jpg")
        )

        if not self.names:
            raise RuntimeError(
                f"Empty cache for split '{split}'. Kindly run build cache firstly('{split}')."
            )

        self.augment = augment

        self.color_jitter = RandomApply(
            [
                ColorJitter(
                    brightness=0.2,
                    contrast=0.2,
                    saturation=0.2
                )
            ],
            p=0.5
        )

    def __len__(self):
        return len(self.names)

    def __getitem__(self, index):

        name = self.names[index]

        image = Image.open(
            self.image_dir / f"{name}.jpg"
        ).convert("RGB")

        mask = Image.open(
            self.mask_dir / f"{name}.png"
        )

        if self.augment:

            if torch.rand(1).item() > 0.5:
                image = TF.hflip(image)
                mask = TF.hflip(mask)

            image = self.color_jitter(image)

        # ImageNet normalization
        image = TF.to_tensor(image)

        # image = TF.normalize(
        #     image,
        #     mean=[0.485, 0.456, 0.406],
        #     std=[0.229, 0.224, 0.225]
        # )

        mask = torch.from_numpy(np.array(mask, dtype=np.uint8))

        return image, mask
```

# SegFormerMLP

```python
class SegFormerMLP(nn.Module):
    def __init__(self, input_dim, embed_dim):
        super().__init__()
        self.proj = nn.Linear(input_dim, embed_dim)

    def forward(self, x):
        # Flatten spatial dimensions: (B, C, H, W) -> (B, C, H*W) -> (B, H*W, C)
        B, C, H, W = x.shape
        x = x.flatten(2).transpose(1, 2)
        x = self.proj(x)
        return x
```

# SegFormerHead

```python
class SegFormerHead(nn.Module):
    def __init__(self, in_channels=[32, 64, 160, 256], embedding_dim=256, num_classes=19):
        super().__init__()

        self.linears = nn.ModuleList([
            SegFormerMLP(c, embedding_dim) for c in in_channels
        ])

        self.linear_fuse = nn.Sequential(
            nn.Conv2d(embedding_dim * 4, embedding_dim, kernel_size=1, bias=False),
            nn.BatchNorm2d(embedding_dim),
            nn.ReLU(inplace=True)
        )

        self.dropout = nn.Dropout2d(0.1)

        self.classifier = nn.Conv2d(embedding_dim, num_classes, kernel_size=1)

    def forward(self, features):
        B = features[0].shape[0]

        out_size = features[0].shape[2:]
        outs = []
        for i, x in enumerate(features):

            x = self.linears[i](x)

            H_i = features[i].shape[2]
            W_i = features[i].shape[3]
            x = x.transpose(1, 2).reshape(B, -1, H_i, W_i)

            if i != 0:
                x = F.interpolate(x, size=out_size, mode='bilinear', align_corners=False)
            outs.append(x)

        out = torch.cat(outs, dim=1) # Shape: (B, embedding_dim * 4, H/4, W/4)

        # Fusion & Classification
        out = self.linear_fuse(out)
        out = self.dropout(out)
        out = self.classifier(out)

        return out
```

# SegformerB0

```python
class SegFormerB0(nn.Module):
    def __init__(self, num_classes=19, pretrained=True):
        super().__init__()

        if pretrained:
            self.backbone = SegformerModel.from_pretrained("/content/mit-b0-local/mit-b0-local")
        else:
            config = SegformerConfig()
            self.backbone = SegformerModel(config)

        self.decode_head = SegFormerHead(
            in_channels=[32, 64, 160, 256],
            embedding_dim=256,
            num_classes=num_classes
        )
        self.register_buffer("mean", torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1))
        self.register_buffer("std", torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1))
    def forward(self, x):
        x = (x - self.mean) / self.std
        input_shape = x.shape[-2:]

        outputs = self.backbone(pixel_values=x, output_hidden_states=True)

        features = outputs.hidden_states

        # Đưa qua Decoder
        out = self.decode_head(features)

        out = F.interpolate(out, size=input_shape, mode='bilinear', align_corners=False)

        return out
```

```python
criterion = nn.CrossEntropyLoss(
    ignore_index=IGNORE_INDEX
)
```

# Confusion Matrix

```python
class SegmentationMetrics:

    def __init__(self, num_classes, device):

        self.num_classes = num_classes

        self.confusion_matrix = torch.zeros(
            (num_classes, num_classes),
            dtype=torch.int64,
            device=device
        )

    @torch.no_grad()
    def update(self, predictions, targets):

        predictions = predictions.argmax(dim=1)

        valid = targets != IGNORE_INDEX

        targets = targets[valid].long()
        predictions = predictions[valid].long()

        indices = (
            targets * self.num_classes + predictions
        )

        matrix = torch.bincount(
            indices,
            minlength=self.num_classes ** 2
        )

        self.confusion_matrix += matrix.reshape(
            self.num_classes,
            self.num_classes
        )

    def compute(self):

        cm = self.confusion_matrix.float()

        tp = torch.diag(cm)

        fp = cm.sum(dim=0) - tp

        fn = cm.sum(dim=1) - tp

        # IoU
        union = tp + fp + fn

        iou = tp / union.clamp(min=1e-8)

        valid_iou = union > 0

        miou = iou[valid_iou].mean()

        # Precision
        precision = tp / (tp + fp).clamp(min=1e-8)

        # Recall
        recall = tp / (tp + fn).clamp(min=1e-8)

        # F1-score
        f1 = (
            2 * precision * recall /
            (precision + recall).clamp(min=1e-8)
        )

        # Pixel accuracy
        pixel_acc = (
            tp.sum() / cm.sum().clamp(min=1e-8)
        )

        return {
            "pixel_accuracy": pixel_acc.item(),
            "mIoU": miou.item(),
            "precision": precision.mean().item(),
            "recall": recall.mean().item(),
            "f1_score": f1.mean().item(),
            "class_iou": iou.cpu().numpy()
        }
```

## Training

```python
def train_one_epoch(
    model,
    loader,
    optimizer,
    criterion,
    scaler,
    device
):
    model.train()

    total_loss = torch.zeros((), device=device)
    correct_pixels = torch.zeros((), device=device)
    total_pixels = torch.zeros((), device=device)
    num_samples = 0

    progress = tqdm(loader, desc="Training")

    for step, (images, masks) in enumerate(progress):

        images = images.to(
            device,
            non_blocking=True
        ).contiguous(memory_format=torch.channels_last)

        masks = masks.to(
            device,
            dtype=torch.long,
            non_blocking=True
        )

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda")
        ):
            outputs = model(images)

            loss = criterion(outputs, masks)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.detach() * images.size(0)
        num_samples += images.size(0)

        with torch.no_grad():
            predictions = outputs.detach().argmax(dim=1)
            valid = masks != IGNORE_INDEX
            correct_pixels += ((predictions == masks) & valid).sum()
            total_pixels += valid.sum()

        if step % 50 == 0:
            progress.set_postfix(loss=f"{loss.item():.4f}")

    avg_loss = (total_loss / num_samples).item()
    accuracy = (correct_pixels.float() / total_pixels.clamp(min=1)).item()

    return avg_loss, accuracy
```

```python
def calculate_metrics(confusion_matrix):

    cm = confusion_matrix.float()

    # True Positive
    TP = torch.diag(cm)

    # False Positive
    FP = cm.sum(dim=0) - TP

    # False Negative
    FN = cm.sum(dim=1) - TP

    # IoU
    iou = TP / (TP + FP + FN).clamp(min=1e-7)

    # Precision
    precision = TP / (TP + FP).clamp(min=1e-7)

    # Recall
    recall = TP / (TP + FN).clamp(min=1e-7)

    # F1-score
    f1 = (
        2 * precision * recall
        / (precision + recall).clamp(min=1e-7)
    )

    # Pixel Accuracy
    pixel_accuracy = (
        TP.sum() / cm.sum().clamp(min=1e-7)
    )

    # Mean IoU
    present_classes = (TP + FP + FN) > 0

    miou = iou[present_classes].mean()

    # Macro Precision / Recall / F1
    precision_mean = precision[present_classes].mean()
    recall_mean = recall[present_classes].mean()
    f1_mean = f1[present_classes].mean()

    return {
        "pixel_accuracy": pixel_accuracy.item(),
        "mIoU": miou.item(),
        "precision": precision_mean.item(),
        "recall": recall_mean.item(),
        "f1_score": f1_mean.item(),
        "class_iou": iou.cpu().numpy()
    }
```

## Validation

```python
def validate(
    model,
    loader,
    criterion,
    device
):
    model.eval()

    total_loss = 0.0
    num_samples = 0

    confusion_matrix = torch.zeros(
        (NUM_CLASSES, NUM_CLASSES),
        device=device,
        dtype=torch.int64
    )

    progress = tqdm(loader, desc="Validation")

    with torch.no_grad():
        for images, masks in progress:
            images = images.to(
                device,
                non_blocking=True
            ).contiguous(memory_format=torch.channels_last)

            masks = masks.to(
                device,
                dtype=torch.long,
                non_blocking=True
            )

            with torch.autocast(
                device_type=device.type,
                dtype=torch.float16,
                enabled=(device.type == "cuda")
            ):
                outputs = model(images)
                loss = criterion(outputs, masks)

            total_loss += loss.item() * images.size(0)
            num_samples += images.size(0)

            predictions = outputs.argmax(dim=1)
            valid = masks != IGNORE_INDEX

            target_pixels = masks[valid]
            predicted_pixels = predictions[valid]

            indices = target_pixels * NUM_CLASSES + predicted_pixels

            confusion_matrix += torch.bincount(
                indices,
                minlength=NUM_CLASSES * NUM_CLASSES
            ).reshape(NUM_CLASSES, NUM_CLASSES)

    avg_loss = total_loss / num_samples
    metrics = calculate_metrics(confusion_matrix)

    return avg_loss, metrics
```

## Main Training

```python
def main():

    # Dataset
    train_dataset = CityscapesDataset(
        split="train",
        augment=True
    )

    val_dataset = CityscapesDataset(
        split="val",
        augment=False
    )

    # DataLoader
    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=(DEVICE.type == "cuda"),
        drop_last=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=(DEVICE.type == "cuda")
    )

    # Model
    model = SegFormerB0(
            num_classes=NUM_CLASSES,
            pretrained=True
        ).to(DEVICE, memory_format=torch.channels_last)

    # Loss
    criterion = nn.CrossEntropyLoss(
        ignore_index=IGNORE_INDEX
    )

    optimizer = torch.optim.AdamW(
        [
            {
                "params": model.backbone.parameters(),
                "lr": BACKBONE_LR
            },
            {
                "params": model.decode_head.parameters(),
                "lr": HEAD_LR
            }
        ],
        weight_decay=WEIGHT_DECAY
    )
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=4
    )

    # Mixed precision
    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=(DEVICE.type == "cuda")
    )

    os.makedirs(
        "checkpoints",
        exist_ok=True
    )

    best_miou = 0.0

    checkpoint_path = "checkpoints/latest_model_var2.pth"
    start_epoch = 0
    best_miou = 0.0

    if os.path.exists(checkpoint_path):
        print(f"Loading checkpoint from {checkpoint_path}...")
        checkpoint = torch.load(checkpoint_path, map_location=DEVICE, weights_only=False)

        model.load_state_dict(checkpoint["model_state_dict"])
        optimizer.load_state_dict(checkpoint["optimizer_state_dict"])

        if "scheduler_state_dict" in checkpoint:
            scheduler.load_state_dict(checkpoint["scheduler_state_dict"])
        if "scaler_state_dict" in checkpoint:
            scaler.load_state_dict(checkpoint["scaler_state_dict"])

        start_epoch = checkpoint["epoch"]
        best_miou = checkpoint.get("best_miou", 0.0)

        print(f"Resumed successfully from epoch {start_epoch} | Best mIoU: {best_miou:.4f}")
    TOTAL_EPOCHS = 50
    for epoch in range(start_epoch, TOTAL_EPOCHS):
        print(f"\nEpoch {epoch + 1}/{TOTAL_EPOCHS}")
        # Training
        train_loss, train_accuracy = train_one_epoch(
            model,
            train_loader,
            optimizer,
            criterion,
            scaler,
            DEVICE
        )

        # Validation
        val_loss, val_metrics = validate(
            model,
            val_loader,
            criterion,
            DEVICE
        )

        scheduler.step(
            val_metrics["mIoU"]
        )

        current_lr_backbone = optimizer.param_groups[0]["lr"]
        current_lr_head = optimizer.param_groups[1]["lr"]

        print("\nTraining Results")
        print(f"Train Loss: {train_loss:.4f}")
        print(f"Train Pixel Accuracy: {train_accuracy:.4f}")

        print("\nValidation Results")
        print(f"Val Loss: {val_loss:.4f}")

        print(
            f"Pixel Accuracy: "
            f"{val_metrics['pixel_accuracy']:.4f}"
        )

        print(
            f"mIoU: {val_metrics['mIoU']:.4f}"
        )

        print(
            f"Precision: {val_metrics['precision']:.4f}"
        )

        print(
            f"Recall: {val_metrics['recall']:.4f}"
        )

        print(
            f"F1-score: {val_metrics['f1_score']:.4f}"
        )

        print(
            f"Backbone LR: {current_lr_backbone:.6f}"
        )

        print(
            f"Head LR: {current_lr_head:.6f}"
        )

        # Save best model
        if val_metrics["mIoU"] > best_miou:
            best_miou = val_metrics["mIoU"]
            torch.save(
                {
                    "epoch": epoch + 1,
                    "model_state_dict": model.state_dict(),
                    "optimizer_state_dict": optimizer.state_dict(),
                    "scheduler_state_dict": scheduler.state_dict(), # Thêm dòng này
                    "scaler_state_dict": scaler.state_dict(),       # Thêm dòng này
                    "best_miou": best_miou,
                    "class_names": CLASS_NAMES
                },
                "/content/best_model_var2.pth"
            )
            print("Best model saved!")

        # Save latest checkpoint
        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "scheduler_state_dict": scheduler.state_dict(), # Thêm dòng này
                "scaler_state_dict": scaler.state_dict(),       # Thêm dòng này
                "best_miou": best_miou
            },
            "/content/latest_model_var2.pth"
        )

    print("\nTraining completed!")
    print(f"Best mIoU: {best_miou:.4f}")


if __name__ == "__main__":
    main()
```

# Inference – Semantic Segmentation

```python
def load_model(checkpoint_path, device):

    model = SegFormerB0(
        num_classes=NUM_CLASSES,
        pretrained=False
    ).to(device)

    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    model.eval()

    return model


@torch.no_grad()
def predict(model, image_path, device):

    image = Image.open(image_path).convert("RGB")

    original_size = image.size

    image = TF.resize(
        image,
        [IMAGE_HEIGHT, IMAGE_WIDTH],
        interpolation=InterpolationMode.BILINEAR
    )

    image = TF.to_tensor(image)

    # image = TF.normalize(
    #     image,
    #     mean=[0.485, 0.456, 0.406],
    #     std=[0.229, 0.224, 0.225]
    # )

    image = image.unsqueeze(0).to(device)

    with torch.autocast(
        device_type=device.type,
        dtype=torch.float16,
        enabled=(device.type == "cuda")
    ):

        output = model(image)

    prediction = output.argmax(dim=1)

    prediction = prediction.squeeze(0).cpu().numpy()

    prediction = Image.fromarray(
        prediction.astype(np.uint8)
    )

    prediction = prediction.resize(
        original_size,
        resample=Image.Resampling.NEAREST
    )

    return prediction
```

```python
CITYSCAPES_COLORS = [
    (128, 64, 128),   # Road
    (244, 35, 232),   # Sidewalk
    (70, 70, 70),     # Building
    (102, 102, 156),  # Wall
    (190, 153, 153),  # Fence
    (153, 153, 153),  # Pole
    (250, 170, 30),   # Traffic light
    (220, 220, 0),    # Traffic sign
    (107, 142, 35),   # Vegetation
    (152, 251, 152),  # Terrain
    (70, 130, 180),   # Sky
    (220, 20, 60),    # Person
    (255, 0, 0),      # Rider
    (0, 0, 142),      # Car
    (0, 0, 70),       # Truck
    (0, 60, 100),     # Bus
    (0, 80, 100),     # Train
    (0, 0, 230),      # Motorcycle
    (119, 11, 32)     # Bicycle
]


def colorize_mask(mask):

    mask = np.array(mask)

    color_mask = np.zeros(
        (*mask.shape, 3),
        dtype=np.uint8
    )

    for class_id, color in enumerate(CITYSCAPES_COLORS):

        color_mask[mask == class_id] = color

    return Image.fromarray(color_mask)
```
